```python
# ===== COMPLETE WORKING CODE FOR COLAB =====

# Step 1: Install required packages
!pip install openpyxl -q

# Step 2: Import libraries
import pandas as pd
import glob
from datetime import datetime
from google.colab import files

# Step 3: Create sample files (skip this if uploading your own)
print("Creating sample files for testing...")
for i in range(1, 4):
    sample = pd.DataFrame({
        'date': ['2024-01-01', '2024-01-02', '2024-01-03'],
        'product': [f'Product A{i}', f'Product B{i}', f'Product C{i}'],
        'sales': [100 * i, 150 * i, 200 * i]
    })
    sample.to_csv(f'sales_file_{i}.csv', index=False)
print("✅ Sample files created!\n")

# Step 4: Define the report function
def clean_sales_data(df):
    """Clean raw sales data"""
    df = df.drop_duplicates()
    if 'date' in df.columns:
        df['date'] = pd.to_datetime(df['date'], errors='coerce')
        df = df.dropna(subset=['date'])
    df = df.fillna(0)
    return df

def generate_report():
    """Main function to generate sales report"""
    all_files = glob.glob("*.csv") + glob.glob("*.xlsx")

    if not all_files:
        print("❌ No files found!")
        print("   Either upload files using files.upload() or create sample files")
        return None

    print(f"Found {len(all_files)} files:")
    for f in all_files:
        print(f"  - {f}")

    # Read all files
    all_data = []
    for file in all_files:
        if file.endswith('.csv'):
            df = pd.read_csv(file)
        else:
            df = pd.read_excel(file)
        all_data.append(df)

    # Combine and clean
    master_df = pd.concat(all_data, ignore_index=True)
    master_df = clean_sales_data(master_df)

    # Calculate summary
    if 'sales' in master_df.columns:
        total_sales = master_df['sales'].sum()
        print(f"\n📊 Total sales: ${total_sales:,.2f}")

    # Save report
    output_file = f"sales_report_{datetime.now().strftime('%Y%m%d_%H%M%S')}.xlsx"
    master_df.to_excel(output_file, index=False)
    print(f"\n✅ Report saved: {output_file}")

    # Download to your computer
    files.download(output_file)

    return output_file

# Step 5: Run it!
generate_report()


```
